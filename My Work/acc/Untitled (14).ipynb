```python
def dremio_query_prod(query, env="prod"):
    with DremioDB(env) as dremio:
        result = dremio.execute_df(query)
    return result


def read_sql(f_name):
    f_path = os.path.join(f_name)
    with open(f_path, "r") as f:
        contents = f.read()
    return contents
```

```python
step_ach = dremio_query_prod(
    read_sql(
        "./sqls/starting_population_autopay_ach.sql"
    )
)
print(f"Length of the step_ach dataset: {len(step_ach)}")

step_two = dremio_query_prod(
    read_sql(
        "./sqls/starting_population_autopay_check_debit.sql"
    )
)
print(f"Length of the step_two dataset: {len(step_two)}")

data = pd.concat([step_ach, step_two])

print(f"Length of the entire original dataset: {len(data)}")

data.customer_id = data.customer_id.astype(int)
data["measurement_date"] = pd.to_datetime(data.measurement_date)
data["run_date"] = pd.to_datetime(data.run_date)
data = data.fillna(0)
```

```python
t = AppendableTable(table='collections_autopay_email_eligible_responsys', org='ops', partner='avant', env='prod')
```

```python
for i in t.list_datasets():
    t.clear_dataset(i)
```

```python
t.upload_df(data, dataset_name='collections_autopay_email_eligible_responsys',overwrite=True)
```
